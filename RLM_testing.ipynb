```python
N = 2
import random
print("--- COMPLETE NETWORK ---")
consensus = []
max_conv_time = 0
for seed in range(10000):
    graph = gen_complete_connected(N)
    game = RepeatedGame(graph=graph, graph_type="complete_connected", rng=np.random.default_rng(seed))
    game.play()


    # print(f"True State: {game.true_state}")
    # print("Action History:")
    # for t, actions in enumerate(game.history):
    #     print(f" t={t}: {actions}")
    metrics = game.convergence_metrics()
    # print(f"\nAbsorbing State Reached: {metrics[0]}")
    consensus.append(int(metrics[1]))
    max_conv_time = max(max_conv_time, metrics[2])
    # print(f"Consensus on True State: {metrics[1]}")
    # print(f"Converged at t={metrics[2]}")
print(np.mean(consensus))
print(max_conv_time)
```
