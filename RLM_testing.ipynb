```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from simulations.game_engine import RepeatedGame, RepeatedBeliefEngine

# 1. RLM Complete Graph Generator (All-to-All)
def gen_rlm_complete(N):
    graph = nx.DiGraph()
    graph.add_nodes_from(range(N))
    edges = [(i, j) for i in range(N) for j in range(N) if i != j]
    graph.add_edges_from(edges)
    return graph

# 2. Run the Game
N = 100
# (Assuming RepeatedGame and RepeatedBeliefEngine are already defined in the cell above)
graph = gen_rlm_complete(N)

print("Starting Repeated Game Simulation (Complete Graph)")
print("-" * 50)

# Instantiate and play
game = RepeatedGame(graph=graph, graph_type="complete", signal_type="unbounded", rng=np.random.default_rng(42))
game.play()

# 3. Extract Metrics
converged, success, lock_time, final_acc = game.convergence_metrics()

print(f"True State of the World: {game.true_state}")
print(f"Converged (Absorbing State Reached)? {converged}")
print(f"Time to Convergence: t = {lock_time}")
print(f"Consensus on True State? {success}")
print(f"Final Accuracy: {final_acc * 100:.1f}%\n")

# 4. Visualize the History
print("Turn-by-Turn Action History:")
print("(Rows = Time steps, Columns = Agents)\n")
history_matrix = np.array(game.history)
print(history_matrix)
```
