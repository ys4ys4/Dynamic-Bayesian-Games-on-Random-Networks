```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from simulations.game_engine import RepeatedGame
from simulations.networks import gen_connected_star, gen_dictator_star

# Assuming your helper functions are in a file called run_experiments.py
from simulations.run_experiments import (
    run_sim, 
    run_overnight_sim, 
    summarise_cascade_metrics
)

# Set plotting style for academic looking graphs
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
N = 5
import random
print("--- CONNECTED STAR ---")
seed = random.randint(0, 10000)
connected_graph = gen_connected_star(N)
connected_game = RepeatedGame(graph=connected_graph, graph_type="connected_star", seed=seed)
connected_game.play()

print(seed)

print(f"True State: {connected_game.true_state}")
print("Action History:")
for t, actions in enumerate(connected_game.history):
    print(f" t={t}: {actions}")

metrics = connected_game.convergence_metrics()
print(f"\nAbsorbing State Reached: {metrics[0]}")
print(f"Consensus on True State: {metrics[1]}")
print(f"Converged at t={metrics[2]}")
```
