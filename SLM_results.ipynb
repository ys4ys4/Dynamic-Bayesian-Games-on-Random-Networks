```python
import os
import re
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from simulations.experiment_configs import ER_BOUNDED_CONFIG, THEOREM_1_CONFIG, THEOREM_2_CONFIG, THEOREM_3I_CONFIG, THEOREM_3II_CONFIG
```

```python
# ER BOUNDED LARGE SIMULATION

config = ER_BOUNDED_CONFIG

summary_df = pd.read_csv(config.output_csv_path)
plot_df = summary_df.copy()
plot_df[config.result_column_name] = plot_df[
    config.result_column_name
].astype(float)
for column in [
    'prob_false_cascade',
    'prob_true_cascade',
    'prob_no_consensus',
    'avg_success_time',
    'avg_fail_time'
]:
    plot_df[column] = pd.to_numeric(plot_df[column], errors='coerce')

fig, (ax_prob, ax_time) = plt.subplots(
    2,
    1,
    figsize=config.plot_figsize,
    sharex=True
)

ax_prob.plot(
    plot_df[config.result_column_name],
    plot_df['prob_false_cascade'],
    marker='o',
    color='red',
    label='False cascade'
)
ax_prob.plot(
    plot_df[config.result_column_name],
    plot_df['prob_true_cascade'],
    marker='o',
    color='green',
    label='True cascade'
)
ax_prob.plot(
    plot_df[config.result_column_name],
    plot_df['prob_no_consensus'],
    marker='o',
    color='gray',
    label='No consensus'
)
ax_prob.set_ylabel('Outcome probability')
ax_prob.set_title(config.plot_title)
ax_prob.legend()
ax_prob.grid(True, alpha=0.3)

ax_time.plot(
    plot_df[config.result_column_name],
    np.log(plot_df['avg_success_time']),
    marker='o',
    color='green',
    label='ln(Average success time)'
)
ax_time.plot(
    plot_df[config.result_column_name],
    np.log(plot_df['avg_fail_time']),
    marker='o',
    color='red',
    label='ln(Average fail time)'
)
ax_time.set_xlabel(config.plot_xlabel)
ax_time.set_ylabel('ln(Mean lock-in time)')
ax_time.legend()
ax_time.grid(True, alpha=0.3)

fig.tight_layout()
fig.savefig(config.plot_path, dpi=config.plot_dpi)
print(f"Loaded data from: {os.path.abspath(config.output_csv_path)}")
print(f"Saved plot to: {os.path.abspath(config.plot_path)}")
plt.show()
```

```python
# THEOREM 1 LARGE SIMULATION

config = THEOREM_1_CONFIG

summary_df = pd.read_csv(config.output_csv_path)

import ast

# Use this for fresh data instead of the regex!
summary_df['final_accuracies'] = summary_df['final_accuracies'].apply(ast.literal_eval)

plot_df = summary_df.explode('final_accuracies', ignore_index=True)
plot_df = plot_df.rename(columns={'final_accuracies': 'Final Accuracy'})

plot_df[config.result_column_name] = plot_df[config.result_column_name].astype(int)
plot_df['Final Accuracy'] = plot_df['Final Accuracy'].astype(float)

plt.figure(figsize=config.plot_figsize)
heatmap = plt.hist2d(
    plot_df[config.result_column_name],
    plot_df['Final Accuracy'],
    bins=config.plot_bins,
    cmap=config.plot_cmap,
    cmin=config.plot_cmin,
)
plt.colorbar(heatmap[3], label='Number of Runs')
plt.title(config.plot_title)
plt.xlabel(config.plot_xlabel)
plt.ylabel(config.plot_ylabel)
plt.tight_layout()
plt.savefig(config.plot_path, dpi=config.plot_dpi)
print(f"Loaded data from: {os.path.abspath(config.output_csv_path)}")
print(f"Saved heatmap to: {os.path.abspath(config.plot_path)}")
plt.show()
```
